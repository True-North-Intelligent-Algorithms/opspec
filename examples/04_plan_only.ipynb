# Test axes processing plans. 

This notebook can be used to test the opspec axes processing plans.

Set the below parameters then run the next cell and check if the plan is correct. 

```processing_axes```: axes to be consumed for processing  
```data_axes```: axes of the data   
```selected position``` : axes name and fixed position   


````python
from opspec.op import Axes, ParamSpec
from opspec.plan import build

processing_axes = list("yxt")
# the op: one parameter, three axes, no function behind it
parameter_spec = ParamSpec(name="image", type=object, default=None, axes=Axes(processing_axes))

# the data: axes and shape, no array behind it
data_axes = list("tczyx")
data_shape = (100, 2, 50, 256, 256)
selected_position = {"z": 5}

# a named axis in selected_position is an axis we sit on, not loop over
dispositions = {data_axes.index(name): "select" for name in selected_position}
````

````python
p = build(parameter_spec, data_axes, data_shape, position=selected_position, dispositions=dispositions)

print(p.summary)
print()
print(f"op {processing_axes} fills data axes {p.mapping}")
print(f"selected {selected_position} becomes (axis, index) {p.select}")
print(f"iterate over data axes {p.iterate}")
print(f"transpose to {p.transpose}")
print(f"output axes {p.output_axes}")
print(f"calls {p.calls}")
print(f"uses_all_data {p.uses_all_data}")
````

Change any of the parameters above and run again.

`uses_all_data` is False when an axis gets discarded: here 50 z planes, one used.
